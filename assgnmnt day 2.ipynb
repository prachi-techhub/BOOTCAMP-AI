```python
import pandas as pd

# Create sample data
data = {
    "Name": ["Rahul", "Priya", "Amit", "Neha"],
    "Email": ["rahul@gmail.com", "priya@yahoo.com", "amit@gmail.com", "invalid-email"],
    "Phone": ["98765-43210", "87654 32109", "76543-21098", "12345"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Clean phone numbers using RegEx
df["Phone"] = df["Phone"].str.replace(r"\D", "", regex=True)

# Validate email using RegEx
df["Valid_Email"] = df["Email"].str.match(
    r"^[\w\.-]+@[\w\.-]+\.\w+$"
)

print("\nCleaned Data:")
print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate values
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Plot sine wave
plt.plot(x, y)

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.grid(True)

plt.show()
```

# Helmet Detection Using YOLO

**Objective:** To detect helmets in images using YOLO and OpenCV and identify whether a helmet is present.

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")
```

```python
results = model(image, conf=0.10, verbose=False)

print("YOLO detection completed!")
print("Total objects detected:", len(results[0].boxes))
```

```python
detected_objects = []

for box in results[0].boxes:
    class_id = int(box.cls[0])
    object_name = model.names[class_id]
    confidence = float(box.conf[0])

    detected_objects.append(object_name)

    print(f"Object: {object_name} | Confidence: {confidence:.2f}")
```

```python
annotated_image = results[0].plot()

annotated_image_rgb = cv2.cvtColor(
    annotated_image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(12, 7))
plt.imshow(annotated_image_rgb)
plt.axis("off")
plt.title("YOLO Detection Result")
plt.show()
```

```python
from collections import Counter

object_counts = Counter(detected_objects)

print("Object Detection Summary")
print("-" * 30)

for object_name, count in object_counts.items():
    print(f"{object_name}: {count}")
```

```python
from huggingface_hub import hf_hub_download

helmet_model_path = hf_hub_download(
    repo_id="sharathhhhh/safetyHelmet-detection-yolov8",
    filename="best.pt"
)

print("Helmet detection model downloaded successfully!")
print(helmet_model_path)
```

```python
helmet_model = YOLO(helmet_model_path)

print("Helmet detection model loaded successfully!")
print("Classes:", helmet_model.names)
```

```python
helmet_results = helmet_model(
    image,
    conf=0.25,
    verbose=False
)

print("Helmet detection completed!")
print("Total detections:", len(helmet_results[0].boxes))
```

```python
helmet_status = []

for box in helmet_results[0].boxes:
    class_id = int(box.cls[0])
    status = helmet_model.names[class_id]
    confidence = float(box.conf[0])

    helmet_status.append(status)

    print(f"Status: {status} | Confidence: {confidence:.2f}")
```

```python
annotated_helmet = helmet_results[0].plot()

annotated_helmet_rgb = cv2.cvtColor(
    annotated_helmet,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(12, 7))
plt.imshow(annotated_helmet_rgb)
plt.axis("off")
plt.title("Helmet Detection Result")
plt.show()
```

```python
from collections import Counter

status_counts = Counter(helmet_status)

with_helmet = status_counts.get("With helmet", 0)
without_helmet = status_counts.get("  Without helmet", 0)

print("=" * 40)
print("       HELMET DETECTION SUMMARY")
print("=" * 40)

print(f"With helmet    : {with_helmet}")
print(f"Without helmet : {without_helmet}")
print(f"Total detected : {with_helmet + without_helmet}")

print("=" * 40)
```

```python
print("=" * 55)
print("             HELMET DETECTION REPORT")
print("=" * 55)

print("\n1. TECHNOLOGY USED")
print("• Python")
print("• OpenCV")
print("• YOLOv8")
print("• Matplotlib")
print("• Custom Helmet Detection Model")

print("\n2. IMAGE ANALYSIS")
print("• Image file: helmet2.0.webp")
print(f"• Total helmet-related detections: {len(helmet_results[0].boxes)}")

print("\n3. HELMET STATUS")
print(f"• With helmet: {with_helmet}")
print(f"• Without helmet: {without_helmet}")

print("\n4. DETECTION DETAILS")

for box in helmet_results[0].boxes:
    class_id = int(box.cls[0])
    status = helmet_model.names[class_id]
    confidence = float(box.conf[0])

    print(f"• {status.strip()} | Confidence: {confidence:.2f}")

print("\n5. OBSERVATION")

if without_helmet > 0:
    print("People without helmets were detected.")
else:
    print("All detected helmet-related objects were classified as wearing helmets.")

print("\n6. OVERALL RESULT")
print("The helmet-specific YOLO model successfully classified")
print("helmet status in the given image.")

print("\n" + "=" * 55)
print("                  REPORT COMPLETE")
print("=" * 55)
```
